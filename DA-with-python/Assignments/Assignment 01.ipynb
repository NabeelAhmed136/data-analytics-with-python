3. Write a program in which use all the operators we can use in Python

```python
 # Arithmetic Operators
num1 = 30
num2 = 6
print(num1 + num2)
print(num1 - num2)
print(num1 / num2)
print(num1 % num2)
print(num1 * num2)
print(num1**num2)

# Assignment operators
x = 5
y = 7
x = y
print(x)
x = 5
x += 3
print(x)
x -= 2
print(x)
x *= 9
print(x)
y = x
print(y)
# Logical opertors ---> and , or , not
maths_marks = 67
physics_marks = 72
print(maths_marks > 70 and physics_marks < 75)
print(maths_marks > 65 and physics_marks > 70)
print(maths_marks < 70 and physics_marks > 90)
# ---> Or operator
print(maths_marks > 65 or physics_marks < 60)
print(maths_marks > 70 or physics_marks > 80)
print(not(maths_marks < 60))
print(not(physics_marks > 65))
# Comparsion Operators
a = 15
b = 25
print(a == b)
print(a > b)
print(a <= b)
print(a != b)
print(a < b)




```

5. Take the values of two variables `a` and `b`, then swap their values without using a third variable and print the result before and after swapping

```python
a = int(input("Enter the value of a: "))
b = int(input("Enter the value of b: "))
print("Before Swapping: ")
print("a= " , a)
print("b= " , b)
# After Swapping
a , b = b , a
print("After Swapping:")
print("a =" , a)
print("b =" , b)

```

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
pi = 3.14
radius = float(input("Enter the radius of the circle: "))
print("radius: " , radius)
area = pi*radius**2
circumference = 2 * pi * radius
print("Area of the circle: " , round(area , 2))
print("Circumference of the circle: " , round(circumference , 2))
```
